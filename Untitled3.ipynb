```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
```

Date needs nothing done to it since it has no missing values.

Time's missing in lots of rows and since it's noth something we actually need for the question then we just fill gaps with "unknown" instead of guessing something never recorded.

Location's missing in only small number of rows, but we need location later for grouping by country and region we just drop the few rows instead of inventing places.

Operator is missing in a small number of rows too. There is no way to guess which airline it was, so those get filled with "Unknown".

Flight number is missing in most of the dataset. Since it barely has any data and we never use it in the analysis, the cleanest option is to drop the whole column.

Route is missing in a lot of rows. It is text and not something the questions ask about directly, so filling it with "Unknown" is enough instead of losing rows over it.

Type of aircraft has a small number of missing values. Fill those with "Unknown" since there is nothing in the other columns that tells us the aircraft type.

Registration is missing in a fair number of rows. It is just an identifier code, so fill with "Unknown".

cn/In (construction and line number) is missing a lot. It is also just an identifier and not used anywhere in the analysis, so fill with "Unknown".

Aboard is missing in a small number of rows. Since this is a number we actually use in the analysis, filling with the median is better than dropping rows, because median is not thrown off by the huge crashes with hundreds of people on board.

Fatalities is missing in a small number of rows too. Same reasoning as Aboard, fill with the median since it is a number we rely on for most of the later questions.

Ground is missing in a small number of rows. A missing value here most likely means nobody on the ground died, so fill those with 0 rather than the median.

Summary is missing in a good number of rows. It is just a text description and not used in any calculation, so fill it with something like "No summary available".

```python
fatality_locations = df[['Date', 'Location', 'Aboard', 'Fatalities']].copy()
fatality_locations.head()
```

```python
highest_fatality_date = fatality_locations.loc[fatality_locations['Fatalities'].idxmax(), 'Date']
print(f"Date with the highest number of fatalities: {highest_fatality_date}")
```

```python
comparison = fatality_locations[['Date', 'Aboard', 'Fatalities']].copy()
comparison['Survivors'] = comparison['Aboard'] - comparison['Fatalities']
comparison.head()
```

```python
no_fatality_crashes = comparison[comparison['Fatalities'] == 0]
print(f"Number of crashes with no fatalities: {len(no_fatality_crashes)}")
no_fatality_crashes.head()
```

```python
fatality_locations[['Region', 'State_Country']] = fatality_locations['Location'].str.split(',', n=1, expand=True)
fatality_locations['Region'] = fatality_locations['Region'].str.strip()
fatality_locations['State_Country'] = fatality_locations['State_Country'].str.strip()
fatality_locations.head()
```

```python
top_100_fatalities = fatality_locations.sort_values(by='Fatalities', ascending=False).head(100)
top_100_fatalities.head()
```

```python
top_25 = top_100_fatalities.nlargest(25, 'Fatalities')

plt.figure(figsize=(9, 9))
colors = sns.color_palette('pastel', len(top_25))
plt.pie(
    top_25['Fatalities'],
    labels=top_25['State_Country'],
    autopct='%1.1f%%',
    colors=colors,
    startangle=90
)
plt.title('Top 25 Fatalities by Country / U.S. State')
plt.axis('equal')
plt.show()
```
